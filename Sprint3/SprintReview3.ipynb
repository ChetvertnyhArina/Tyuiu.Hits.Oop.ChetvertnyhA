```polyglot-notebook
public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return "Название: " + Name + ", Цена: " + Price + " руб.";
    }

    public override string ToString()
    {
        return GetInfo();
    }

    public static bool operator ==(Component a, Component b)
    {
        if (ReferenceEquals(a, b))
            return true;
        if (a is null || b is null)
            return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is Component other)
        {
            return this == other;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}


public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return "CPU: " + Name + ", ядер: " + Cores + ", цена: " + Price + " руб.";
    }

    public void Install()
    {
        Console.WriteLine("  -> Процессор " + Name + " установлен в сокет материнской платы");
    }
}


public class RAM : Component, IInstallable
{
    public int SizeGB { get; set; }

    public RAM(string name, decimal price, int sizeGB) : base(name, price)
    {
        SizeGB = sizeGB;
    }

    public override string GetInfo()
    {
        return "RAM: " + Name + ", объем: " + SizeGB + " ГБ, цена: " + Price + " руб.";
    }

    public void Install()
    {
        Console.WriteLine("  -> Оперативная память " + Name + " установлена в слот DIMM");
    }
}


public class GPU : Component, IInstallable
{
    public int MemoryGB { get; set; }

    public GPU(string name, decimal price, int memoryGB) : base(name, price)
    {
        MemoryGB = memoryGB;
    }

    public override string GetInfo()
    {
        return "GPU: " + Name + ", видеопамять: " + MemoryGB + " ГБ, цена: " + Price + " руб.";
    }

    public void Install()
    {
        Console.WriteLine("  -> Видеокарта " + Name + " установлена в слот PCIe");
    }
}


public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }

    public HDD(string name, decimal price, int capacityGB) : base(name, price)
    {
        CapacityGB = capacityGB;
    }

    public override string GetInfo()
    {
        return "HDD: " + Name + ", емкость: " + CapacityGB + " ГБ, цена: " + Price + " руб.";
    }

    public void Install()
    {
        Console.WriteLine("  -> Жесткий диск " + Name + " подключен по интерфейсу SATA");
    }
}


public class Computer
{
    
    private List<Component> components = new List<Component>();

   
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        Console.WriteLine("Добавлен компонент: " + component.GetInfo());
        component.Install();
    }

    
    public void RemoveComponent<T>(T component) where T : Component
    {
        bool removed = components.Remove(component);
        if (removed)
        {
            Console.WriteLine("Удален компонент: " + component.Name);
        }
        else
        {
            Console.WriteLine("Компонент " + component.Name + " не найден в списке");
        }
    }

    
    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (Component comp in components)
        {
            total = total + comp.Price;
        }
        return total;
    }

   
    public void ShowInfo()
    {
        Console.WriteLine("\nИнформация о компьютере");
        Console.WriteLine("Количество компонентов: " + components.Count);
        foreach (Component comp in components)
        {
            Console.WriteLine("  - " + comp.GetInfo());
        }
        Console.WriteLine("Общая стоимость: " + GetTotalPrice() + " руб.");
    }
}


public static class ComputerFactory
{
    public static Computer CreateComputer<T>(List<T> componentList) where T : Component, IInstallable
    {
        Computer computer = new Computer();
        foreach (T comp in componentList)
        {
            computer.AddComponent(comp);
        }
        return computer;
    }

    
    public static Computer CreateGamingPC()
    {
        Console.WriteLine("Создаем игровой компьютер");
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i9", 45000, 16));
        pc.AddComponent(new RAM("Kingston Fury", 18000, 32));
        pc.AddComponent(new GPU("NVIDIA RTX 4080", 95000, 16));
        pc.AddComponent(new HDD("Samsung 980 Pro", 12000, 1000));
        return pc;
    }
}



Console.WriteLine("Создание компонентов");


CPU myCpu = new CPU("Intel Core i7", 30000, 8);
RAM myRam = new RAM("Corsair Vengeance", 12000, 16);
GPU myGpu = new GPU("NVIDIA RTX 3070", 60000, 8);
HDD myHdd = new HDD("Seagate Barracuda", 5000, 1000);


Console.WriteLine(myCpu.GetInfo());
Console.WriteLine(myRam.GetInfo());
Console.WriteLine(myGpu.GetInfo());
Console.WriteLine(myHdd.GetInfo());


Console.WriteLine("\nДемонстрация перегрузки ToString():");
Console.WriteLine(myCpu.ToString());

Console.WriteLine("\nПерегрузка операторов == и !=");


CPU cpu1 = new CPU("Intel Core i7", 30000, 8);
CPU cpu2 = new CPU("Intel Core i7", 30000, 8);
CPU cpu3 = new CPU("AMD Ryzen 7", 28000, 8);

Console.WriteLine("cpu1 == cpu2 (одинаковые): " + (cpu1 == cpu2));
Console.WriteLine("cpu1 != cpu3 (разные): " + (cpu1 != cpu3));

Console.WriteLine("\nСоздаем компьютер вручную");

Computer firstComputer = new Computer();
firstComputer.AddComponent(myCpu);
firstComputer.AddComponent(myRam);
firstComputer.AddComponent(myGpu);
firstComputer.AddComponent(myHdd);

firstComputer.ShowInfo();

Console.WriteLine("\nУдаление компонента из первого компьютера");

firstComputer.RemoveComponent(myHdd);
firstComputer.ShowInfo();





Console.WriteLine("\nСоздаем компьютер через фабрику");
Computer secondComputer = ComputerFactory.CreateGamingPC();
secondComputer.ShowInfo();

Console.WriteLine("\nGeneric метод CreateComputer<T>");

List<CPU> cpuList = new List<CPU>();
cpuList.Add(new CPU("AMD Ryzen 9", 40000, 12));

Computer thirdComputer = ComputerFactory.CreateComputer(cpuList);
thirdComputer.ShowInfo();

Console.WriteLine("\nДемонстрация полиморфизма");


Component[] allComponents = new Component[]
{
    new CPU("Intel Core i5", 20000, 6),
    new RAM("Kingston HyperX", 8000, 16),
    new GPU("AMD RX 6700", 45000, 12),
    new HDD("Toshiba P300", 4000, 2000)
};

Console.WriteLine("Полиморфный вывод информации о компонентах:");
foreach (Component comp in allComponents)
{
   
    Console.WriteLine("  " + comp.GetInfo());
}
```
