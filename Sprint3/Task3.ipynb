Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма. Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой. Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование.

Требования к базовому классу Project:

• Атрибуты: ID проекта (ProjectId), Название проекта (ProjectName), Статус
проекта (Status).

• Методы:

o UpdateStatus(): метод для обновления статуса проекта.

o GetProjectDetails(): метод для получения деталей проекта.

o AssignTeamMember(): метод для назначения командного члена
проекту.

Требования к производным классам:
1. ПродуктовыйПроект (ProductProject): Должен содержать дополнительные
атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate).
Метод UpdateStatus() должен быть переопределен для включения
ожидаемой даты завершения в статус проекта.
2. ИсследовательскийПроект (ResearchProject): Должен содержать
дополнительные атрибуты, такие как Финансирование (FundingAmount).
Метод AssignTeamMember() должен быть переопределен для указания
специфических требований к навыкам командного члена для
исследовательского проекта.
3. ИнфраструктурныйПроект (InfrastructureProject) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как Срок
выполнения (ExecutionPeriod). Метод GetProjectDetails() должен быть
переопределен для отображения срока выполнения проекта вместе с
другими деталями проекта.

```
public interface IReportable {
    void GenerateReport();
}

public interface IFinancial {
    decimal Budget { get; }
    void SpendMoney(decimal amount);
}

public class TeamMember {
    public string Name { get; set; }
    public string Skill { get; set; }
    public int ExperienceYears { get; set; }

    public TeamMember(string name, string skill = "Не указан", int experienceYears = 0) {
        Name = name;
        Skill = skill;
        ExperienceYears = experienceYears;
    }

    public void Promote() {
        ExperienceYears++;
        Console.WriteLine($"[TeamMember] {Name} повышен! Опыт: {ExperienceYears} лет.");
    }

    public override string ToString() => $"{Name} ({Skill}, стаж: {ExperienceYears} лет)";
}

public class Project {
    public int ProjectId { get; set; }
    public string ProjectName { get; set; }
    public string Status { get; set; }
    public string Manager { get; set; }
    public decimal Budget { get; set; }
    public List<TeamMember> Team { get; set; } = new List<TeamMember>();

    public Project(int id, string name, string status, string manager = "Не назначен", decimal budget = 0) {
        ProjectId = id;
        ProjectName = name;
        Status = status;
        Manager = manager;
        Budget = budget;
    }

    
    public virtual void UpdateStatus(string newStatus) {
        Status = newStatus;
        Console.WriteLine($"[Base] Статус '{ProjectName}' изменен на '{Status}'");
    }


    public virtual void UpdateStatus(string newStatus, string reason) {
        Status = newStatus;
        Console.WriteLine($"[Base] Статус '{ProjectName}' -> '{Status}'. Причина: {reason}");
    }


    public virtual string GetProjectDetails() {
        return $"ID:{ProjectId} | {ProjectName} | Статус:{Status} | Бюджет:{Budget} | Менеджер:{Manager}";
    }


    public virtual string GetProjectDetails(bool showTeam) {
        string details = GetProjectDetails();
        if (showTeam && Team.Count > 0) {
            details += " | Команда: " + string.Join(", ", Team.Select(m => m.Name));
        }
        return details;
    }

  
    public virtual void AssignTeamMember(TeamMember member) {
        Team.Add(member);
        Console.WriteLine($"[Base] {member.Name} добавлен в проект '{ProjectName}'");
    }

    public virtual void AssignTeamMember(TeamMember member, string role) {
        Team.Add(member);
        Console.WriteLine($"[Base] {member.Name} назначен на роль '{role}' в '{ProjectName}'");
    }
}


public class ProductProject : Project {
    public DateTime ExpectedCompletionDate { get; set; }
    public string Version { get; set; }

    public ProductProject(int id, string name, string status, DateTime date, string version = "1.0") 
        : base(id, name, status) {
        ExpectedCompletionDate = date;
        Version = version;
    }


    public override void UpdateStatus(string newStatus) {
        base.UpdateStatus(newStatus);
        Console.WriteLine($"[Product] Ожидаемая дата завершения: {ExpectedCompletionDate:d}");
    }

    public override void UpdateStatus(string newStatus, string reason) {
        base.UpdateStatus(newStatus, reason);
        Console.WriteLine($"[Product] Ожидаемая дата завершения: {ExpectedCompletionDate:d}");
    }

    public override string GetProjectDetails() {
        return base.GetProjectDetails() + $" | Версия:{Version} | Срок:{ExpectedCompletionDate:d}";
    }

    public void ReleaseVersion(string newVersion) {
        Version = newVersion;
        Console.WriteLine($"[Product] Выпущена версия {newVersion} для '{ProjectName}'");
    }
}


public class ResearchProject : Project {
    public decimal FundingAmount { get; set; }
    public string Topic { get; set; }

    public ResearchProject(int id, string name, string status, decimal funding, string topic) 
        : base(id, name, status) {
        FundingAmount = funding;
        Topic = topic;
    }

  
    public override void AssignTeamMember(TeamMember member) {
        if (!member.Skill.Contains("PhD") && !member.Skill.Contains("Ученый")) {
            Console.WriteLine($"[Research] Ошибка: {member.Name} не имеет степени PhD для проекта '{Topic}'");
            return;
        }
        base.AssignTeamMember(member);
    }

    public override void AssignTeamMember(TeamMember member, string role) {
        if (!member.Skill.Contains("PhD") && !member.Skill.Contains("Ученый")) {
            Console.WriteLine($"[Research] Ошибка: {member.Name} не имеет степени PhD для проекта '{Topic}'");
            return;
        }
        base.AssignTeamMember(member, role);
    }

    public override string GetProjectDetails() {
        return base.GetProjectDetails() + $" | Тема:{Topic} | Финансирование:{FundingAmount}";
    }
}


public class AppliedResearchProject : ResearchProject, IReportable {
    public string IndustryPartner { get; set; }

    public AppliedResearchProject(int id, string name, string status, decimal funding, string topic, string partner) 
        : base(id, name, status, funding, topic) {
        IndustryPartner = partner;
    }

   
    public void GenerateReport() {
        Console.WriteLine($"[AppliedResearch] Отчет по исследованию '{ProjectName}' сгенерирован. Партнер: {IndustryPartner}");
    }

    public override string GetProjectDetails() {
        return base.GetProjectDetails() + $" | Партнер: {IndustryPartner}";
    }
}


public class InfrastructureProject : Project, IReportable, IFinancial {
    public string ExecutionPeriod { get; set; }
    public string Location { get; set; }

    public InfrastructureProject(int id, string name, string status, string period, string location, decimal budget) 
        : base(id, name, status, budget: budget) {
        ExecutionPeriod = period;
        Location = location;
    }

    
    public override string GetProjectDetails() {
        return base.GetProjectDetails() + $" | Срок: {ExecutionPeriod} | Локация: {Location}";
    }

    
    public void GenerateReport() => Console.WriteLine($"[Infra] Отчет по инфраструктуре '{ProjectName}' сгенерирован.");
    public void SpendMoney(decimal amount) {
        Budget -= amount;
        Console.WriteLine($"[Infra] Потрачено {amount}. Остаток бюджета: {Budget}");
    }
}


public class Organization {
    public string Name { get; set; }
    public string Ceo { get; set; }
    private List<Project> _projects = new List<Project>();

    public Organization(string name, string ceo = "Не указан") {
        Name = name;
        Ceo = ceo;
    }

    public void AddProject(Project project) {
        _projects.Add(project);
        Console.WriteLine($"[Organization] Проект '{project.ProjectName}' добавлен в '{Name}'");
    }

    public void AssignMemberToProject(int projectId, TeamMember member, string role = null) {
        var project = _projects.FirstOrDefault(p => p.ProjectId == projectId);
        if (project == null) {
            Console.WriteLine($"[Organization] Проект с ID={projectId} не найден.");
            return;
        }
        if (role != null) project.AssignTeamMember(member, role);
        else project.AssignTeamMember(member);
    }

    public void PrintSummary() {
        Console.WriteLine($"\nСводка по организации '{Name}' (CEO: {Ceo})");
        foreach (var p in _projects)
            Console.WriteLine($" {p.GetProjectDetails(true)}");
    }
}


public class ProjectRegistry<T> where T : Project {
    private List<T> _projects = new List<T>();

    public void Add(T project) {
        _projects.Add(project);
        Console.WriteLine($"[Registry] Проект '{project.ProjectName}' зарегистрирован.");
    }

    public void PrintByStatus(string status) {
        Console.WriteLine($"\n--- Поиск по статусу '{status}' ---");
        var found = _projects.Where(p => p.Status == status).ToList();
        if (!found.Any()) Console.WriteLine("Проектов не найдено.");
        else found.ForEach(p => Console.WriteLine(p.GetProjectDetails()));
    }
}


Console.WriteLine("1. Простое наследование (ProductProject)");
var prod = new ProductProject(1, "Мобильное приложение", "В работе", DateTime.Now.AddMonths(6), "1.0");
prod.UpdateStatus("Тестирование"); 
prod.UpdateStatus("Завершен", "Все тесты пройдены"); 
prod.ReleaseVersion("2.0");
Console.WriteLine(prod.GetProjectDetails(true)); 

Console.WriteLine("\n2. Сложное наследование (AppliedResearchProject)");
var research = new AppliedResearchProject(2, "ИИ в медицине", "Исследование", 1000000, "Нейросети", "МедТех");
research.AssignTeamMember(new TeamMember("Сидоров", "Разработчик")); 
research.AssignTeamMember(new TeamMember("Сидоров PhD", "Ученый PhD")); 
research.AssignTeamMember(new TeamMember("Иванова PhD", "Ученый PhD"), "Ведущий исследователь"); 
Console.WriteLine(research.GetProjectDetails());

Console.WriteLine("\n3. Множественное наследование (InfrastructureProject)");
var infra = new InfrastructureProject(3, "Дата-центр", "Строительство", "1 год", "Москва", 5000000);
infra.SpendMoney(1000000);
infra.GenerateReport();
Console.WriteLine(infra.GetProjectDetails());

Console.WriteLine("\n4. Взаимодействие объектов через Organization");
var org = new Organization("Инновационные технологии", "Петров П.П.");
org.AddProject(prod);
org.AddProject(research);
org.AddProject(infra);

org.AssignMemberToProject(1, new TeamMember("Петров", "Тестировщик", 3), "QA Lead");
org.AssignMemberToProject(2, new TeamMember("Смирнова PhD", "Ученый PhD", 10));
org.AssignMemberToProject(3, new TeamMember("Васильева", "Системный администратор", 7));

org.PrintSummary();

Console.WriteLine("\n5. Полиморфизм и Generic-класс");
ProjectRegistry<Project> registry = new ProjectRegistry<Project>();
registry.Add(prod);
registry.Add(research);
registry.Add(infra);

registry.PrintByStatus("В работе");
registry.PrintByStatus("Завершен");

Console.WriteLine("\n6. Полиморфизм через массив интерфейсов");
IReportable[] reporters = { research, infra };
foreach (var r in reporters) r.GenerateReport();
```
