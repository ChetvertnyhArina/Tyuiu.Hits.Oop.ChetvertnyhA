```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветёт.");
    }

    public void Grow(float amount)
    {
        Height += amount;
        Console.WriteLine($"Растение {Name} выросло на {amount} см. Текущая высота: {Height} см.");
    }
}



public class Flower : Plant
{
    public string Color { get; set; }        
    public int PetalCount { get; set; }      

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

  
    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }


    public void ShowInfo()
    {
        Console.WriteLine($"  Цвет: {Color}");
        Console.WriteLine($"  Количество лепестков: {PetalCount}");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}



public class Tree : Plant
{
    public int Age { get; set; }              
    public string LeafType { get; set; }      

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }


    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }


    public void ShowInfo()
    {
        Console.WriteLine($"  Возраст: {Age} лет");
        Console.WriteLine($"  Тип листьев: {LeafType}");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} постарело на {years} лет. Новый возраст: {Age} лет.");
    }
}



Flower rose = new Flower("Роза", 35f, "Белый", 25);
Tree maple = new Tree("Клён", 450f, 12, "Лиственные");


Console.WriteLine("=== Демонстрация методов Розы ===");
rose.Bloom();          
rose.ShowInfo();       
rose.Grow(8f);
rose.Wither();


Console.WriteLine("\n=== Демонстрация методов Клёна ===");
maple.Bloom();        
maple.ShowInfo();      
maple.Grow(25f);
maple.GrowOlder(3);


Console.WriteLine("\n=== Полиморфизм ===");
Plant[] garden = { rose, maple };
foreach (Plant p in garden)
{
    p.Bloom();  
}
```
