Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма. обавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой.

Требования к базовому классу Project:

• Атрибуты: ID проекта (ProjectId), Название проекта (ProjectName), Статус
проекта (Status).

• Методы:

o UpdateStatus(): метод для обновления статуса проекта.

o GetProjectDetails(): метод для получения деталей проекта.

o AssignTeamMember(): метод для назначения командного члена
проекту.

Требования к производным классам:
1. ПродуктовыйПроект (ProductProject): Должен содержать дополнительные
атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate).
Метод UpdateStatus() должен быть переопределен для включения
ожидаемой даты завершения в статус проекта.
2. ИсследовательскийПроект (ResearchProject): Должен содержать
дополнительные атрибуты, такие как Финансирование (FundingAmount).
Метод AssignTeamMember() должен быть переопределен для указания
специфических требований к навыкам командного члена для
исследовательского проекта.
3. ИнфраструктурныйПроект (InfrastructureProject) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как Срок
выполнения (ExecutionPeriod). Метод GetProjectDetails() должен быть
переопределен для отображения срока выполнения проекта вместе с
другими деталями проекта.

```
using System;
using System.Collections.Generic;
using System.Linq;


// ИНТЕРФЕЙСЫ для множественного наследования
public interface IBudgetable
{
    decimal Budget { get; }
    decimal SpentAmount { get; }
    void AddExpense(decimal amount, string description);
    string GetBudgetReport();
}

public interface IReportable
{
    int ReportCount { get; }
    void GenerateReport(string reportType);
    string GetLastReport();
}


// БАЗОВЫЙ КЛАСС TeamMember
public class TeamMember
{
    private string _name;
    private string _skill;
    private string _email;
    private int _experienceYears;

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя участника не может быть пустым");
            _name = value;
        }
    }

    public string Skill
    {
        get => _skill;
        set => _skill = value ?? "Не указан";
    }

    public string Email
    {
        get => _email;
        set => _email = string.IsNullOrWhiteSpace(value) ? "не указан" : value;
    }

    public int ExperienceYears
    {
        get => _experienceYears;
        set => _experienceYears = value >= 0 ? value : 0;
    }

    public TeamMember(string name, string skill = "Не указан", 
                      string email = "", int experienceYears = 0)
    {
        Name = name;
        Skill = skill;
        Email = email;
        ExperienceYears = experienceYears;
    }

    public void Promote()
    {
        ExperienceYears++;
        Console.WriteLine($"[TeamMember] {Name} повышен! Опыт: {ExperienceYears} лет.");
    }

    public override string ToString() => $"{Name} ({Skill}, стаж: {ExperienceYears} лет)";
}

// БАЗОВЫЙ КЛАСС Project
public class Project
{
    private int _projectId;
    private string _projectName;
    private string _status;
    private string _priority;
    private string _description;
    private DateTime _startDate;

    public int ProjectId
    {
        get => _projectId;
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть положительным");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get => _projectName;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым");
            _projectName = value;
        }
    }

    public string Status
    {
        get => _status;
        set => _status = value ?? "Не определён";
    }

    public string Priority
    {
        get => _priority;
        set => _priority = value ?? "Средний";
    }

    public string Description
    {
        get => _description;
        set => _description = value ?? "Нет описания";
    }

    public DateTime StartDate
    {
        get => _startDate;
        set => _startDate = value;
    }

    protected List<TeamMember> TeamMembers = new List<TeamMember>();

    public Project(int projectId, string projectName, string status, 
                   string priority = "Средний", string description = "", 
                   DateTime? startDate = null)
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Status = status;
        Priority = priority;
        Description = description;
        StartDate = startDate ?? DateTime.Today;
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Project] Статус проекта '{ProjectName}' обновлён на: {Status}");
    }

    public virtual string GetProjectDetails()
    {
        return $"ID: {ProjectId}, Название: {ProjectName}, Статус: {Status}, " +
               $"Приоритет: {Priority}, Участников: {TeamMembers.Count}, " +
               $"Старт: {StartDate:dd.MM.yyyy}";
    }

    public virtual void AssignTeamMember(TeamMember member)
    {
        TeamMembers.Add(member);
        Console.WriteLine($"[Project] Участник '{member.Name}' добавлен в проект '{ProjectName}'");
    }

    public void RemoveTeamMember(string memberName)
    {
        var member = TeamMembers.FirstOrDefault(m => m.Name == memberName);
        if (member != null)
        {
            TeamMembers.Remove(member);
            Console.WriteLine($"[Project] Участник '{memberName}' удалён из проекта '{ProjectName}'");
        }
    }

    public int GetTeamSize() => TeamMembers.Count;

    public IReadOnlyList<TeamMember> GetTeamMembers() => TeamMembers.AsReadOnly();
}


// ПРОЕКТ 1: ProductProject
// ПРОСТОЕ НАСЛЕДОВАНИЕ: ProductProject -> Project
public class ProductProject : Project
{
    private DateTime _expectedCompletionDate;
    private string _version;
    private string _platform;

    public DateTime ExpectedCompletionDate
    {
        get => _expectedCompletionDate;
        set => _expectedCompletionDate = value;
    }

    public string Version
    {
        get => _version;
        set => _version = value ?? "1.0";
    }

    public string Platform
    {
        get => _platform;
        set => _platform = value ?? "Не указана";
    }

    public ProductProject(int projectId, string projectName, string status,
                          DateTime expectedCompletionDate,
                          string version = "1.0", string platform = "Не указана",
                          string priority = "Средний", string description = "")
        : base(projectId, projectName, status, priority, description)
    {
        ExpectedCompletionDate = expectedCompletionDate;
        Version = version;
        Platform = platform;
    }

    public override void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[ProductProject] Статус продукта '{ProjectName}' (v{Version}) обновлён: " +
                          $"{Status}. Ожидаемая дата: {ExpectedCompletionDate:dd.MM.yyyy}");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Версия: {Version}, Платформа: {Platform}, " +
               $"Ожидаемая дата: {ExpectedCompletionDate:dd.MM.yyyy}";
    }

    public void ReleaseVersion(string newVersion)
    {
        Version = newVersion;
        Console.WriteLine($"[ProductProject] Выпущена версия {newVersion} продукта '{ProjectName}'");
    }
}


// ПРОЕКТ 2: ResearchProject
// СЛОЖНОЕ НАСЛЕДОВАНИЕ: AppliedResearchProject -> ResearchProject -> Project
public class ResearchProject : Project
{
    private decimal _fundingAmount;
    private string _topic;
    private bool _publicationRequired;

    public decimal FundingAmount
    {
        get => _fundingAmount;
        set
        {
            if (value < 0)
                throw new ArgumentException("Финансирование не может быть отрицательным");
            _fundingAmount = value;
        }
    }

    public string Topic
    {
        get => _topic;
        set => _topic = value ?? "Не указана";
    }

    public bool PublicationRequired
    {
        get => _publicationRequired;
        set => _publicationRequired = value;
    }

    public ResearchProject(int projectId, string projectName, string status,
                           decimal fundingAmount,
                           string topic = "", bool publicationRequired = false,
                           string priority = "Высокий", string description = "")
        : base(projectId, projectName, status, priority, description)
    {
        FundingAmount = fundingAmount;
        Topic = topic;
        PublicationRequired = publicationRequired;
    }

    public override void AssignTeamMember(TeamMember member)
    {
        string requiredSkill = "PhD в области исследований";
        if (!string.Equals(member.Skill, requiredSkill, StringComparison.OrdinalIgnoreCase))
        {
            Console.WriteLine($"[ResearchProject] Участник '{member.Name}' не соответствует " +
                              $"требованиям (нужен навык: {requiredSkill}).");
            return;
        }
        TeamMembers.Add(member);
        Console.WriteLine($"[ResearchProject] Участник '{member.Name}' добавлен в проект '{ProjectName}'. " +
                          $"Тема: {Topic}. Финансирование: {FundingAmount:C}");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Тема: {Topic}, Финансирование: {FundingAmount:C}, " +
               $"Публикация: {(PublicationRequired ? "обязательна" : "не требуется")}";
    }

    public void PublishResults(string journalName)
    {
        if (PublicationRequired)
            Console.WriteLine($"[ResearchProject] Результаты исследования '{ProjectName}' " +
                              $"опубликованы в журнале '{journalName}'");
        else
            Console.WriteLine($"[ResearchProject] Публикация для '{ProjectName}' не требуется.");
    }
}


// СЛОЖНОЕ НАСЛЕДОВАНИЕ + МНОЖЕСТВЕННОЕ
// AppliedResearchProject -> ResearchProject -> Project
// + реализует IReportable
public class AppliedResearchProject : ResearchProject, IReportable
{
    private int _reportCount;
    private int _riskLevel;
    private string _lastReport;   // НОВОЕ: для интерфейса IReportable

    public int ReportCount
    {
        get => _reportCount;
        set => _reportCount = value >= 0 ? value : 0;
    }

    public int RiskLevel
    {
        get => _riskLevel;
        set => _riskLevel = (value >= 0 && value <= 10) ? value : 5;
    }

    // От интерфейса IReportable
    public string GetLastReport() => _lastReport ?? "Отчётов нет";

    public AppliedResearchProject(int projectId, string projectName, string status,
                                   decimal fundingAmount, string topic,
                                   bool publicationRequired = true,
                                   string priority = "Высокий", string description = "")
        : base(projectId, projectName, status, fundingAmount, topic, publicationRequired, priority, description)
    {
        ReportCount = 0;
        RiskLevel = 0;
    }

    // От интерфейса IReportable
    public void GenerateReport(string reportType)
    {
        ReportCount++;
        _lastReport = $"{reportType} от {DateTime.Today:dd.MM.yyyy}";
        Console.WriteLine($"[AppliedResearchProject] Сформирован отчёт '{reportType}' " +
                          $"по проекту '{ProjectName}'. Всего отчётов: {ReportCount}");
    }

    public void AddRisk(string riskDescription, int severity)
    {
        RiskLevel = Math.Min(10, RiskLevel + severity);
        Console.WriteLine($"[AppliedResearchProject] Добавлен риск: {riskDescription}. " +
                          $"Общий уровень риска: {RiskLevel}/10");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Отчётов: {ReportCount}, Уровень риска: {RiskLevel}/10";
    }
}

// ПРОЕКТ 3: InfrastructureProject
// МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: InfrastructureProject -> Project + IBudgetable + IReportable
public class InfrastructureProject : Project, IBudgetable, IReportable
{
    private string _executionPeriod;
    private string _location;
    private int _criticality;
    private decimal _budget;
    private decimal _spentAmount;
    private int _reportCount;
    private string _lastReport;

    public string ExecutionPeriod
    {
        get => _executionPeriod;
        set => _executionPeriod = value ?? "Не указан";
    }

    public string Location
    {
        get => _location;
        set => _location = value ?? "Не указано";
    }

    public int Criticality
    {
        get => _criticality;
        set => _criticality = (value >= 1 && value <= 10) ? value : 5;
    }

    // От интерфейса IBudgetable
    public decimal Budget
    {
        get => _budget;
        set => _budget = value >= 0 ? value : 0;
    }

    public decimal SpentAmount
    {
        get => _spentAmount;
        private set => _spentAmount = value;
    }

    // От интерфейса IReportable
    public int ReportCount
    {
        get => _reportCount;
        private set => _reportCount = value;
    }

    public string GetLastReport() => _lastReport ?? "Отчётов нет";

    public InfrastructureProject(int projectId, string projectName, string status,
                                 string executionPeriod,
                                 string location = "", int criticality = 5,
                                 decimal budget = 0,
                                 string priority = "Высокий", string description = "")
        : base(projectId, projectName, status, priority, description)
    {
        ExecutionPeriod = executionPeriod;
        Location = location;
        Criticality = criticality;
        Budget = budget;
        SpentAmount = 0;
        ReportCount = 0;
    }

    // От интерфейса IBudgetable
    public void AddExpense(decimal amount, string description)
    {
        if (SpentAmount + amount > Budget)
        {
            Console.WriteLine($"[InfrastructureProject] Превышение бюджета '{ProjectName}'! " +
                              $"Нельзя потратить {amount:C}");
            return;
        }
        SpentAmount += amount;
        Console.WriteLine($"[InfrastructureProject] Расход {amount:C} на '{description}'. " +
                          $"Остаток: {Budget - SpentAmount:C}");
    }

    public string GetBudgetReport()
    {
        return $"Бюджет '{ProjectName}': {Budget:C}, потрачено: {SpentAmount:C}, " +
               $"остаток: {Budget - SpentAmount:C}";
    }

    // От интерфейса IReportable
    public void GenerateReport(string reportType)
    {
        ReportCount++;
        _lastReport = $"{reportType} от {DateTime.Today:dd.MM.yyyy}";
        Console.WriteLine($"[InfrastructureProject] Сформирован отчёт '{reportType}' " +
                          $"по проекту '{ProjectName}'. Всего отчётов: {ReportCount}");
    }

    public void UpgradeCriticality(int newLevel)
    {
        Criticality = newLevel;
        Console.WriteLine($"[InfrastructureProject] Критичность '{ProjectName}' изменена на {newLevel}/10");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Срок: {ExecutionPeriod}, Локация: {Location}, " +
               $"Критичность: {Criticality}/10, Бюджет: {Budget:C}, Отчётов: {ReportCount}";
    }
}


// Organization
public class Organization
{
    private string _name;
    private List<Project> _projects = new List<Project>();
    private string _ceo;

    public string Name
    {
        get => _name;
        set => _name = value ?? "Без названия";
    }

    public string Ceo
    {
        get => _ceo;
        set => _ceo = value ?? "Не указан";
    }

    public Organization(string name, string ceo = "Не указан")
    {
        Name = name;
        Ceo = ceo;
    }

    public void AddProject(Project project)
    {
        _projects.Add(project);
        Console.WriteLine($"[Organization] Проект '{project.ProjectName}' добавлен в '{Name}'");
    }

    public Project FindProject(int projectId) =>
        _projects.FirstOrDefault(p => p.ProjectId == projectId);

    public void AssignMemberToProject(int projectId, TeamMember member)
    {
        var project = FindProject(projectId);
        if (project == null)
        {
            Console.WriteLine($"[Organization] Проект с ID={projectId} не найден.");
            return;
        }
        project.AssignTeamMember(member);
    }

    public void TransferMember(int fromProjectId, int toProjectId, string memberName)
    {
        var from = FindProject(fromProjectId);
        var to = FindProject(toProjectId);
        if (from == null || to == null)
        {
            Console.WriteLine("[Organization] Один из проектов не найден.");
            return;
        }
        var member = from.GetTeamMembers().FirstOrDefault(m => m.Name == memberName);
        if (member == null)
        {
            Console.WriteLine($"[Organization] Участник '{memberName}' не найден в проекте {fromProjectId}.");
            return;
        }
        from.RemoveTeamMember(memberName);
        to.AssignTeamMember(member);
        Console.WriteLine($"[Organization] {memberName} переведён из проекта {fromProjectId} в {toProjectId}");
    }

    public void PrintSummary()
    {
        Console.WriteLine($"\nСводка по организации '{Name}' (CEO: {Ceo})");
        foreach (var p in _projects)
            Console.WriteLine($" {p.GetProjectDetails()}");
    }
}


var today = DateTime.Today;
var org = new Organization("Инновационные технологии", "Петров П.П.");

Console.WriteLine("ПРОЕКТ 1: ProductProject");
Console.WriteLine("ПРОСТОЕ НАСЛЕДОВАНИЕ: ProductProject -> Project");
var productProject = new ProductProject(1, "Мобильное приложение", "В разработке",
                                         today.AddMonths(6), "1.0", "iOS/Android",
                                         "Высокий", "Разработка мобильного приложения");
Console.WriteLine(productProject.GetProjectDetails());
productProject.ReleaseVersion("2.0");

Console.WriteLine("\nПРОЕКТ 2: AppliedResearchProject");
Console.WriteLine("СЛОЖНОЕ НАСЛЕДОВАНИЕ: AppliedResearchProject -> ResearchProject -> Project");
var researchProject = new AppliedResearchProject(2, "ИИ для медицины", "Исследование",
                                                  1_500_000m, "Нейросети в диагностике",
                                                  true, "Высокий", "Исследование ИИ");
Console.WriteLine(researchProject.GetProjectDetails());
researchProject.GenerateReport("Промежуточный");
researchProject.GenerateReport("Финальный");
researchProject.AddRisk("Задержка оборудования", 3);
researchProject.PublishResults("Nature");

Console.WriteLine("\nПРОЕКТ 3: InfrastructureProject");
Console.WriteLine("МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: InfrastructureProject -> Project + IBudgetable + IReportable");
var infraProject = new InfrastructureProject(3, "Дата-центр", "Планирование",
                                              $"{today:dd.MM.yyyy} - {today.AddYears(1):dd.MM.yyyy}",
                                              "Франкфурт", 9, 10_000_000m,
                                              "Критический", "Облачная инфраструктура");
Console.WriteLine(infraProject.GetProjectDetails());
infraProject.AddExpense(2_000_000m, "Серверы");
infraProject.AddExpense(3_000_000m, "Сетевое оборудование");
infraProject.GenerateReport("Аудит безопасности");
infraProject.UpgradeCriticality(10);
Console.WriteLine(infraProject.GetBudgetReport());

Console.WriteLine("\nВзаимодействие через Organization");
org.AddProject(productProject);
org.AddProject(researchProject);
org.AddProject(infraProject);

Console.WriteLine("\nНазначение участников");
org.AssignMemberToProject(1, new TeamMember("Смирнова А.", "Разработчик", "smirnova@mail.ru", 5));
org.AssignMemberToProject(1, new TeamMember("Кузнецов К.", "Тестировщик", "kuznetsov@mail.ru", 3));
org.AssignMemberToProject(2, new TeamMember("Иванов И.", "PhD в области исследований", "ivanov@mail.ru", 8));
org.AssignMemberToProject(3, new TeamMember("Васильева Е.", "Системный администратор", "vasilieva@mail.ru", 7));

org.PrintSummary();

Console.WriteLine("\nПолиморфизм через интерфейсы");
Console.WriteLine("•Проекты с управлением бюджетом (IBudgetable)");
IBudgetable[] budgetables = { infraProject };
foreach (var b in budgetables)
    Console.WriteLine(" " + b.GetBudgetReport());

Console.WriteLine("•Проекты с поддержкой отчётности (IReportable)");
IReportable[] reportables = { researchProject, infraProject };
foreach (var r in reportables)
    r.GenerateReport("Ежемесячный");
```
