<h1 style="color:DodgerBlue">Сложное наследование</h1>

Сложное наследование в C# отличается от простого тем, что оно позволяет создавать более глубокие иерархии классов, где один класс может наследоваться от другого, который уже является производным классом. Это позволяет еще больше абстрагировать общие свойства и методы, делая код более модульным и легким для поддержки.

Давайте рассмотрим пример сложного наследования, используя классы `Vehicle`, `Car`, и `Motorcycle`, а также введем новый класс `SportsCar`, который будет наследоваться от `Car`.

### Шаг 1: Базовый класс Vehicle

Начнем с определения базового класса `Vehicle`, как и в предыдущем примере.

```csharp
public class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }

    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}
```

### Шаг 2: Производный класс Car

Теперь определим класс `Car`, который наследуется от `Vehicle`.

```csharp
public class Car : Vehicle
{
    public int Doors { get; set; }

    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}
```

### Шаг 3: Производный класс SportsCar

Введем класс `SportsCar`, который наследуется от `Car`, демонстрируя сложное наследование.

```csharp
public class SportsCar : Car
{
    public bool IsConvertible { get; set; }

    public SportsCar(string make, string model, int doors, bool isConvertible) 
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}
```

### Шаг 4: Производный класс Motorcycle

Определим класс `Motorcycle`, который также наследуется от `Vehicle`.

```csharp
public class Motorcycle : Vehicle
{
    public string EngineType { get; set; }

    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}
```

### Шаг 5: Использование классов

Теперь мы можем создать экземпляры наших классов и использовать их.

```csharp
Car myCar = new Car("Toyota", "Corolla", 4);
myCar.DisplayInfo();

SportsCar mySportsCar = new SportsCar("Ferrari", "488 Spider", 2, true);
mySportsCar.DisplayInfo();

Motorcycle myMotorcycle = new Motorcycle("Ducati", "Panigale V4", "V4");
myMotorcycle.DisplayInfo();
```

### Объяснение кода

- **Базовый класс `Vehicle`** предоставляет основные свойства и методы для всех транспортных средств.
- **Класс `Car`** наследует от `Vehicle` и добавляет свойство `Doors`.
- **Класс `SportsCar`** наследует от `Car`, демонстрируя сложное наследование, и добавляет свойство `IsConvertible`.
- **Класс `Motorcycle`** также наследует от `Vehicle`, но имеет свое собственное свойство `EngineType`.

### Отличие сложного наследования от простого

- **Глубина иерархии**: Сложное наследование позволяет создавать более глубокие иерархии классов, где классы могут наследоваться друг от друга на несколько уровней вниз.
- **Переиспользование кода**: Сложное наследование увеличивает возможности для переиспользования кода, поскольку классы могут наследовать не только непосредственно от базового класса, но и от других производных классов.
- **Флексибельность**: Сложное наследование обеспечивает большую гибкость в проектировании программ, позволяя разработчикам точно определять отношения между классами и их роли в системе.

Таким образом, сложное наследование в C# представляет собой мощный инструмент для создания хорошо структурированных и масштабируемых приложений, позволяя эффективно организовывать код и избегать дублирования.

````
public class Vehicle
{
    // Свойство для хранения марки транспортного средства
    public string Make { get; set; }

    // Свойство для хранения модели транспортного средства
    public string Model { get; set; }

    // Конструктор класса Vehicle
    public Vehicle(string make, string model)
    {
        Make = make;
        Model = model;
    }

    // Виртуальный метод для отображения информации о транспортном средстве
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Производитель: {Make}, Модель: {Model}");
    }
}

public class Car : Vehicle
{
    // Свойство для хранения количества дверей
    public int Doors { get; set; }

    // Конструктор класса Car
    public Car(string make, string model, int doors) : base(make, model)
    {
        Doors = doors;
    }

    // Переопределенный метод DisplayInfo для отображения информации об автомобиле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Двери: {Doors}");
    }
}

// СЛОЖНОЕ наследование: SportsCar -> Car -> Vehicle
public class SportsCar : Car
{
    // Свойство для хранения признака откидного верха
    public bool IsConvertible { get; set; }

    // Конструктор класса SportsCar
    public SportsCar(string make, string model, int doors, bool isConvertible) 
        : base(make, model, doors)
    {
        IsConvertible = isConvertible;
    }

    // Переопределенный метод DisplayInfo для отображения информации о спорткаре
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Автомобиль с откидным верхом: {IsConvertible}");
    }
}

public class Motorcycle : Vehicle
{
    // Свойство для хранения типа двигателя
    public string EngineType { get; set; }

    // Конструктор класса Motorcycle
    public Motorcycle(string make, string model, string engineType) : base(make, model)
    {
        EngineType = engineType;
    }

    // Переопределенный метод DisplayInfo для отображения информации о мотоцикле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип двигателя: {EngineType}");
    }
}

// СЛОЖНОЕ наследование: SportMotorcycle -> Motorcycle -> Vehicle
public class SportMotorcycle : Motorcycle
{
    // Свойство для хранения гоночного класса
    public string RacingClass { get; set; }

    // Конструктор класса SportMotorcycle
    public SportMotorcycle(string make, string model, string engineType, string racingClass) 
        : base(make, model, engineType)
    {
        RacingClass = racingClass;
    }

    // Переопределенный метод DisplayInfo для отображения информации о спортивном мотоцикле
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Гоночный класс: {RacingClass}");
    }
}

// === Использование классов ===

// Простое наследование: Car -> Vehicle
Car myCar = new Car("Mercedes-Benz", "E-Class", 4);
Console.WriteLine("Автомобиль");
myCar.DisplayInfo();

Console.WriteLine();

// Сложное наследование: SportsCar -> Car -> Vehicle
SportsCar mySportsCar = new SportsCar("Porsche", "911 Turbo S", 2, false);
Console.WriteLine("Спорткар");
mySportsCar.DisplayInfo();

Console.WriteLine();


// Сложное наследование: SportMotorcycle -> Motorcycle -> Vehicle
SportMotorcycle mySportMotorcycle = new SportMotorcycle("Ducati", "Panigale V4 R", "V4 90°", "WorldSBK");
Console.WriteLine("Спортивный мотоцикл");
mySportMotorcycle.DisplayInfo();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также сложное наследование.

````
public class Animal
{
    private string _name;
    private int _age;
    private double _weight;

    // Свойство Name с геттером и сеттером
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    // Свойство Age с валидацией в сеттере
    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
        }
    }

    // Свойство Weight с валидацией в сеттере
    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value > 0)
                _weight = value;
        }
    }

    // Статический атрибут — подсчёт всех созданных животных
    public static int TotalAnimals { get; private set; } = 0;

    // Конструктор по умолчанию
    public Animal()
    {
        _name = "Неизвестно";
        _age = 0;
        _weight = 0;
        TotalAnimals++;
    }

    // Конструктор с параметрами
    public Animal(string name, int age, double weight) : this()
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    // Обычный метод
    public void Grow(int years)
    {
        Age += years;
    }

    // Виртуальный метод для отображения информации
    public virtual string Info()
    {
        return $"Животное. Имя: {Name}. Возраст: {Age} лет. Вес: {Weight} кг.";
    }

    // Виртуальный метод передвижения
    public virtual void Move()
    {
        Console.WriteLine($"{Name} двигается.");
    }

    // Статический метод
    public static void DisplayTotalAnimals()
    {
        Console.WriteLine($"Всего животных создано: {TotalAnimals}");
    }
}

// ПРОМЕЖУТОЧНЫЙ класс: Dog -> Animal
public class Dog : Animal
{
    private string _breed;

    public string Breed
    {
        get { return _breed; }
        set { _breed = value; }
    }

    public Dog() : base()
    {
        Breed = "Неизвестна";
    }

    public Dog(string name, int age, double weight, string breed) : base(name, age, weight)
    {
        Breed = breed;
    }

    public void Bark()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} бежит.");
    }

    public override string Info()
    {
        return base.Info() + $" Порода: {Breed}.";
    }
}

// ПРОМЕЖУТОЧНЫЙ класс: Cat -> Animal
public class Cat : Animal
{
    private string _color;

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public Cat() : base()
    {
        Color = "Не указан";
    }

    public Cat(string name, int age, double weight, string color) : base(name, age, weight)
    {
        Color = color;
    }

    public void Meow()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} крадётся.");
    }

    public override string Info()
    {
        return base.Info() + $" Цвет: {Color}.";
    }
}

// ПРОМЕЖУТОЧНЫЙ класс: Bird -> Animal
public class Bird : Animal
{
    private bool _canFly;

    public bool CanFly
    {
        get { return _canFly; }
        set { _canFly = value; }
    }

    public Bird() : base()
    {
        CanFly = true;
    }

    public Bird(string name, int age, double weight, bool canFly) : base(name, age, weight)
    {
        CanFly = canFly;
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит!");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} машет крыльями.");
    }

    public override string Info()
    {
        return base.Info() + $" Умеет летать: {CanFly}.";
    }
}

// СЛОЖНОЕ наследование: GuardDog -> Dog -> Animal
public class GuardDog : Dog
{
    private string _guardedArea;

    public string GuardedArea
    {
        get { return _guardedArea; }
        set { _guardedArea = value; }
    }

    public GuardDog() : base()
    {
        GuardedArea = "Не указана";
    }

    public GuardDog(string name, int age, double weight, string breed, string guardedArea) 
        : base(name, age, weight, breed)
    {
        GuardedArea = guardedArea;
    }

    public void Patrol()
    {
        Console.WriteLine($"{Name} патрулирует территорию: {GuardedArea}.");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} обходит периметр.");
    }

    public override string Info()
    {
        return base.Info() + $" Охраняемая территория: {GuardedArea}.";
    }
}

// СЛОЖНОЕ наследование: PersianCat -> Cat -> Animal
public class PersianCat : Cat
{
    private bool _isLonghaired;

    public bool IsLonghaired
    {
        get { return _isLonghaired; }
        set { _isLonghaired = value; }
    }

    public PersianCat() : base()
    {
        IsLonghaired = true;
    }

    public PersianCat(string name, int age, double weight, string color, bool isLonghaired) 
        : base(name, age, weight, color)
    {
        IsLonghaired = isLonghaired;
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мур-мур-мур!");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} грациозно ступает.");
    }

    public override string Info()
    {
        return base.Info() + $" Длинношёрстный: {IsLonghaired}.";
    }
}

// СЛОЖНОЕ наследование: Penguin -> Bird -> Animal
public class Penguin : Bird
{
    private string _habitat;

    public string Habitat
    {
        get { return _habitat; }
        set { _habitat = value; }
    }

    public Penguin() : base()
    {
        CanFly = false;
        Habitat = "Не указан";
    }

    public Penguin(string name, int age, double weight, string habitat) 
        : base(name, age, weight, false)
    {
        Habitat = habitat;
    }

    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в воде.");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} неуклюже переваливается.");
    }

    public override string Info()
    {
        return base.Info() + $" Среда обитания: {Habitat}.";
    }
}

// === Использование классов  ===

// Сложное наследование: GuardDog -> Dog -> Animal
GuardDog myGuardDog = new GuardDog("Рекс", 4, 25.0, "Немецкая овчарка", "Склад №7");
Console.WriteLine("Сторожевая собака");
Console.WriteLine($"Имя: {myGuardDog.Name}");
Console.WriteLine($"Возраст: {myGuardDog.Age} года");
Console.WriteLine($"Вес: {myGuardDog.Weight} кг");
Console.WriteLine($"Порода: {myGuardDog.Breed}");
Console.WriteLine($"Охраняемая территория: {myGuardDog.GuardedArea}");
myGuardDog.Bark();
myGuardDog.Patrol();
myGuardDog.Move();

Console.WriteLine();

// Сложное наследование: PersianCat -> Cat -> Animal
PersianCat myPersianCat = new PersianCat("Снежок", 3, 4.5, "Белый", true);
Console.WriteLine("Персидский кот");
Console.WriteLine($"Имя: {myPersianCat.Name}");
Console.WriteLine($"Возраст: {myPersianCat.Age} года");
Console.WriteLine($"Вес: {myPersianCat.Weight} кг");
Console.WriteLine($"Цвет: {myPersianCat.Color}");
Console.WriteLine($"Длинношёрстный: {myPersianCat.IsLonghaired}");
myPersianCat.Meow();
myPersianCat.Purr();
myPersianCat.Move();

Console.WriteLine();

// Сложное наследование: Penguin -> Bird -> Animal
Penguin myPenguin = new Penguin("Тукс", 5, 15.0, "Антарктида");
Console.WriteLine("Пингвин");
Console.WriteLine($"Имя: {myPenguin.Name}");
Console.WriteLine($"Возраст: {myPenguin.Age} лет");
Console.WriteLine($"Вес: {myPenguin.Weight} кг");
Console.WriteLine($"Умеет летать: {myPenguin.CanFly}");
Console.WriteLine($"Среда обитания: {myPenguin.Habitat}");
myPenguin.Fly();
myPenguin.Swim();
myPenguin.Move();

Console.WriteLine();
````
