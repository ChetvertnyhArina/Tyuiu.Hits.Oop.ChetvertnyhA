<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Project:

• Атрибуты: ID проекта (ProjectId), Название проекта (ProjectName), Статус
проекта (Status).

• Методы:

o UpdateStatus(): метод для обновления статуса проекта.

o GetProjectDetails(): метод для получения деталей проекта.

o AssignTeamMember(): метод для назначения командного члена
проекту.

Требования к производным классам:
1. ПродуктовыйПроект (ProductProject): Должен содержать дополнительные
атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate).
Метод UpdateStatus() должен быть переопределен для включения
ожидаемой даты завершения в статус проекта.
2. ИсследовательскийПроект (ResearchProject): Должен содержать
дополнительные атрибуты, такие как Финансирование (FundingAmount).
Метод AssignTeamMember() должен быть переопределен для указания
специфических требований к навыкам командного члена для
исследовательского проекта.
3. ИнфраструктурныйПроект (InfrastructureProject) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как Срок
выполнения (ExecutionPeriod). Метод GetProjectDetails() должен быть
переопределен для отображения срока выполнения проекта вместе с
другими деталями проекта.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
using System;
using System.Collections.Generic;
using System.Linq;


public class TeamMember
{
    private string _name;
    private string _skill;

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя участника не может быть пустым");
            _name = value;
        }
    }

    public string Skill
    {
        get => _skill;
        set => _skill = value ?? "Не указан";
    }


    public TeamMember(string name, string skill = "Не указан")
    {
        Name = name;    
        Skill = skill;  
    }

    public override string ToString() => $"{Name} ({Skill})";
}

public class Project
{
    private int _projectId;
    private string _projectName;
    private string _status;

    public int ProjectId
    {
        get => _projectId;
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть положительным");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get => _projectName;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым");
            _projectName = value;
        }
    }

    public string Status
    {
        get => _status;
        set => _status = value ?? "Не определён";
    }

    protected List<TeamMember> TeamMembers = new List<TeamMember>();


    public Project(int projectId, string projectName, string status)
    {
        ProjectId = projectId;      
        ProjectName = projectName;  
        Status = status;            
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Project] Статус проекта '{ProjectName}' обновлён на: {Status}");
    }

    public virtual string GetProjectDetails()
    {
        return $"ID: {ProjectId}, Название: {ProjectName}, Статус: {Status}, " +
               $"Участников: {TeamMembers.Count}";
    }

    public virtual void AssignTeamMember(TeamMember member)
    {
        TeamMembers.Add(member);
        Console.WriteLine($"[Project] Участник '{member.Name}' добавлен в проект '{ProjectName}'");
    }

    
    public IReadOnlyList<TeamMember> GetTeamMembers() => TeamMembers.AsReadOnly();
}


public class ProductProject : Project
{
    private DateTime _expectedCompletionDate;

    public DateTime ExpectedCompletionDate
    {
        get => _expectedCompletionDate;
        set => _expectedCompletionDate = value;
    }


    public ProductProject(int projectId, string projectName, string status,
                          DateTime expectedCompletionDate)
        : base(projectId, projectName, status)
    {
        ExpectedCompletionDate = expectedCompletionDate; 
    }

    public override void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[ProductProject] Статус продукта '{ProjectName}' обновлён: " +
                          $"{Status}. Ожидаемая дата завершения: " +
                          $"{ExpectedCompletionDate:dd.MM.yyyy}");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Ожидаемая дата завершения: {ExpectedCompletionDate:dd.MM.yyyy}";
    }
}


public class ResearchProject : Project
{
    private decimal _fundingAmount;

    public decimal FundingAmount
    {
        get => _fundingAmount;
        set
        {
            if (value < 0)
                throw new ArgumentException("Финансирование не может быть отрицательным");
            _fundingAmount = value;
        }
    }


    public ResearchProject(int projectId, string projectName, string status,
                           decimal fundingAmount)
        : base(projectId, projectName, status)
    {
        FundingAmount = fundingAmount;
    }

    public override void AssignTeamMember(TeamMember member)
    {
        string requiredSkill = "PhD в области исследований";
      
        if (!string.Equals(member.Skill, requiredSkill, StringComparison.OrdinalIgnoreCase))
        {
            Console.WriteLine($"[ResearchProject] Участник '{member.Name}' не соответствует " +
                              $"требованиям (нужен навык: {requiredSkill}).");
            return;
        }

        TeamMembers.Add(member);
        Console.WriteLine($"[ResearchProject] Участник '{member.Name}' добавлен в " +
                          $"исследовательский проект '{ProjectName}'. " +
                          $"Требуемый навык: {requiredSkill}. " +
                          $"Финансирование: {FundingAmount:C}");
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Финансирование: {FundingAmount:C}";
    }
}


public class InfrastructureProject : Project
{
    private string _executionPeriod;

    public string ExecutionPeriod
    {
        get => _executionPeriod;
        set => _executionPeriod = value ?? "Не указан";
    }

    public InfrastructureProject(int projectId, string projectName, string status,
                                 string executionPeriod)
        : base(projectId, projectName, status)
    {
        ExecutionPeriod = executionPeriod; 
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() +
               $", Срок выполнения: {ExecutionPeriod}";
    }
}


public class Organization
{
    private string _name;
    private List<Project> _projects = new List<Project>();

    public string Name
    {
        get => _name;
        set => _name = value ?? "Без названия";
    }

    public Organization(string name)
    {
        Name = name; 
    }

    public void AddProject(Project project)
    {
        _projects.Add(project);
        Console.WriteLine($"[Organization] Проект '{project.ProjectName}' добавлен в организацию '{Name}'");
    }

    public Project FindProject(int projectId)
    {
        return _projects.FirstOrDefault(p => p.ProjectId == projectId);
    }

   
    public void AssignMemberToProject(int projectId, TeamMember member)
    {
        var project = FindProject(projectId);
        if (project == null)
        {
            Console.WriteLine($"[Organization] Проект с ID={projectId} не найден.");
            return;
        }
        project.AssignTeamMember(member); 
    }

    public void PrintSummary()
    {
        Console.WriteLine($"\n=== Сводка по организации '{Name}' ===");
        foreach (var p in _projects)
            Console.WriteLine($" {p.GetProjectDetails()}");
    }
}




var today = DateTime.Today;


var org = new Organization("Инновационные технологии");


var productProject = new ProductProject(1, "Мобильное приложение", "В разработке", today.AddMonths(6));
var researchProject = new ResearchProject(2, "ИИ для медицины", "Исследование", 1_500_000m);
var infraProject = new InfrastructureProject(3, "Дата-центр", "Планирование",
    $"{today:dd.MM.yyyy} - {today.AddYears(1):dd.MM.yyyy}");


org.AddProject(productProject);
org.AddProject(researchProject);
org.AddProject(infraProject);

Console.WriteLine("\n=== Детали проектов ===");
foreach (var p in new Project[] { productProject, researchProject, infraProject })
    Console.WriteLine($" {p.GetProjectDetails()}\n");

Console.WriteLine("=== Обновление статусов ===");
foreach (var p in new Project[] { productProject, researchProject, infraProject })
    p.UpdateStatus("Активный");

Console.WriteLine("\n=== Назначение участников через организацию ===");

org.AssignMemberToProject(1, new TeamMember("Смирнова Анна", "Разработчик"));
org.AssignMemberToProject(1, new TeamMember("Кузнецов Константин", "Тестировщик"));
org.AssignMemberToProject(2, new TeamMember("Иванов Иван", "PhD в области исследований"));
org.AssignMemberToProject(2, new TeamMember("Сидоров Тимур", "Бакалавр")); 
org.AssignMemberToProject(3, new TeamMember("Васильева Елизавета", "Системный администратор"));
org.AssignMemberToProject(3, new TeamMember("Суворов Дмитрий", "Инженер"));


org.PrintSummary();


Console.WriteLine("\n=== Участники проекта 'Мобильное приложение' ===");
foreach (var m in productProject.GetTeamMembers())
    Console.WriteLine($"  - {m}");
```
