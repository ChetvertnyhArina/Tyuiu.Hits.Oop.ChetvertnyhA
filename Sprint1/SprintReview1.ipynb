```polyglot-notebook
public class Product
{

    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;


    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value >= 0 ? value : 0; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value >= 0 ? value : 0; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

   
    public Product(string name, decimal price, int quantity,
                   string category, string description)
    {
        this.name = name;
        this.price = price >= 0 ? price : 0;
        this.quantity = quantity >= 0 ? quantity : 0;
        this.category = category;
        this.description = description;
    }

    
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
        if (quantity < 0) quantity = 0;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}\n" +
               $"  Категория: {category}\n" +
               $"  Описание: {description}\n" +
               $"  Цена: {price:C}\n" +
               $"  Количество: {quantity}\n" +
               $"  Общая стоимость: {GetTotalPrice():C}";
    }
}


public class Seller
{
    
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    
    private List<Product> products = new List<Product>();

    
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value >= 0 ? value : 0; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public List<Product> Products
    {
        get { return products; }
    }

    
    public Seller(string name, string employeeId,
                  decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary >= 0 ? salary : 0;
        this.contactInfo = contactInfo;
    }

    
    public void AddProduct(Product product)
    {
        if (product == null)
        {
            Console.WriteLine("Ошибка: товар не может быть null.");
            return;
        }
        products.Add(product);
        Console.WriteLine($"Товар \"{product.Name}\" добавлен продавцу {name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product == null)
        {
            Console.WriteLine("Ошибка: товар не найден.");
            return;
        }
        if (quantity <= 0)
        {
            Console.WriteLine("Ошибка: количество должно быть положительным.");
            return;
        }
        if (product.Quantity < quantity)
        {
            Console.WriteLine($"Недостаточно товара \"{product.Name}\" на складе. " +
                              $"Доступно: {product.Quantity}.");
            return;
        }

        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продано {quantity} шт. товара \"{product.Name}\". " +
                          $"Остаток: {product.Quantity}.");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name}\n" +
               $"  ID сотрудника: {employeeId}\n" +
               $"  Зарплата: {salary:C}\n" +
               $"  Контакты: {contactInfo}\n" +
               $"  Кол-во товаров: {products.Count}";
    }
}



public class Store
{
    
    private string storeName;
    private string location;
    public string storeHours;

    
    private List<Seller> sellers = new List<Seller>();

    
    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public List<Seller> Sellers
    {
        get { return sellers; }
    }

    
    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    
    public void AddSeller(Seller seller)
    {
        if (seller == null)
        {
            Console.WriteLine("Ошибка: продавец не может быть null.");
            return;
        }
        sellers.Add(seller);
        Console.WriteLine($"Продавец \"{seller.Name}\" добавлен в магазин \"{storeName}\".");
    }

    public void ListProducts()
    {
        Console.WriteLine($"=== Товары магазина \"{storeName}\" ===");
        bool any = false;
        foreach (var seller in sellers)
        {
            foreach (var product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
                Console.WriteLine(new string('-', 40));
                any = true;
            }
        }
        if (!any)
        {
            Console.WriteLine("Товары отсутствуют.");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}\n" +
               $"  Локация: {location}\n" +
               $"  Часы работы: {storeHours}\n" +
               $"  Кол-во продавцов: {sellers.Count}";
    }
}


var product = new Product(
    name: "Кольцо с изумрудом",
    price: 84500m,
    quantity: 4,
    category: "Кольца",
    description: "Золото 585 пробы, изумруд 0.42 карата, огранка «маркиз»"
);
Console.WriteLine(product.GetProductInfo());
Console.WriteLine();


var seller = new Seller(
    name: "Виктория Лазурская",
    employeeId: "JW-118",
    salary: 78000m,
    contactInfo: "v.lazurskaya@aurum-gold.ru"
);
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine();


var store = new Store(
    storeName: "Аурум",
    location: "г. Санкт-Петербург, Невский проспект, 28",
    storeHours: "10:00 - 21:00"
);
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine();


Console.WriteLine("=== Действия ===");
store.AddSeller(seller);
seller.AddProduct(product);
Console.WriteLine();

seller.SellProduct(product, 1);
Console.WriteLine();

product.UpdateQuantity(3);
Console.WriteLine($"После пополнения остаток: {product.Quantity}");
Console.WriteLine();

Console.WriteLine("=== Товары магазина ===");
store.ListProducts();
```
